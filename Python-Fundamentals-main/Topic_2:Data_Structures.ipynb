When you use assignment (=), no new object is created; you’re just making a reference to the same memory, meaning any change affects both variables. A shallow copy (like copy.copy() or list()) creates a new outer object, but the inner elements are still shared references, so changing nested data impacts the original. In contrast, a deep copy (copy.deepcopy()) creates a completely independent clone by recursively duplicating both the outer and all inner objects. While deep copying provides total independence, it has the highest memory usage, whereas assignment is the most memory-efficient.

```python
original = [1, 2, [3, 4]]
alias = original
alias[0] = 'X'
print(alias)
print(original)
alias.append(5)
print(alias)
print(original)
# original is now ['X', 2, [3, 4]]
```

```python
import copy
original = [1, 2, [3, 4]]
shallow = copy.copy(original)
shallow[2][0] = 'X'
print(shallow)
print(original)
shallow.append(5)
print(shallow)
print(original)
#adding will not affect the original but modifying can affect the original
# original is now [1, 2, ['X', 4]] (Nested data changed)
```

```python
import copy
original = [1, 2, [3, 4]]
deep = copy.deepcopy(original)
deep[2][0] = 'X'
print(deep)
print(original)
deep.append(5)
print(deep)
print(original)
# original remains [1, 2, [3, 4]] (Isolated)
```

That is correct. Tuples in Python are immutable data structures, meaning their contents cannot be changed after creation [1]. Operations like concatenation using + or augmented assignment += work around this by creating a new tuple object in memory that holds all the combined elements [1].
Here is a breakdown of how it works:
Original Tuple: When you define tup1 = (1, 2), Python creates a tuple object at a specific memory location.
Operation: When you execute tup2 = tup1 + (3, 4) or tup1 += (3, 4), Python does not alter the original object [1]. Instead, it builds an entirely new tuple object containing (1, 2, 3, 4).
Reassignment: The variable name (tup1 in the augmented assignment example) is then updated to reference this newly created object, while the original (1, 2) object is discarded (and eventually garbage collected) [1].
This behavior ensures data integrity, as the original tuple's state is preserved, aligning with Python's tuple documentation.

New Object Creation for Operations: Operations like concatenation (using the + operator) or augmented assignment (+=) appear to modify the tuple, but in reality, they create a new tuple object in memory with the combined elements, and the variable is then reassigned to this new object. The original tuple remains unchanged.

```python
#1. Basic Function + Parameters You want to calculate the total cost of all items.

prices = [50, 30, 10, 40]
def sum_prices(prices):
  return sum(prices)
sum_prices(prices)
```

```python
#2. Multiple Parameters : You want to display item names with their prices.

items = ["milk", "bread", "eggs", "butter"]
prices = [50, 30, 10, 40]
def together(items,prices):
  for i,p in zip(items,prices): #if you do not use zip it wont be able to traverse both items and pieces in "for" loop hence we use zip to kind of merge items and pieces into a single entity
    print(i,p)
together(items,prices)
```

```python
  #let us say that you do not want to use inbuilt iter rather you want to create your own iterator, then below is the method you would follow
#note that any time you create a iterator of your own it would have an init method, a iter method and a next method with the next having stopiteration exception present

import random

class Dice:
    def __init__(self, rolls): #an init method return object itself, it keep the count of rolls, how many times we would like to roll
        self.rolls = rolls
        self.count = 0

    def __iter__(self): #used to implicitly call the method behind the scenes
        return self

    def __next__(self): #used for incremental purposes
        if self.count < self.rolls: #if the current roll count is less than the roll we asked for
            self.count+=1 #increment the count to 1
            return random.randint(1, 6) #as it is a die then the output must be between 1-6
        else:
            raise StopIteration # very important because otherwise if it goes more than 3 then it would keep printing out none and not stop, hence stopiteration error is a must.
                                #you can try yourself when you remove the else condition the output is "none"

dice = Dice(3)

for die in dice:
    print(die)
```

```python
def bubble_sort(array):
    n = len(array)

    for i in range(n):
        # Create a flag that will allow the function to
        # terminate early if there's nothing left to sort
        already_sorted = True

        # Start looking at each item of the list one by one,
        # comparing it with its adjacent value. With each
        # iteration, the portion of the array that you look at
        # shrinks because the remaining items have already been
        # sorted.
        for j in range(n - i - 1):
            if array[j] > array[j + 1]:
                # If the item you're looking at is greater than its
                # adjacent value, then swap them
                array[j], array[j + 1] = array[j + 1], array[j]

                # Since you had to swap two elements,
                # set the `already_sorted` flag to `False` so the
                # algorithm doesn't finish prematurely
                already_sorted = False

        # If there were no swaps during the last iteration,
        # the array is already sorted, and you can terminate
        if already_sorted:
            break

    return array
```

```python
import asyncio
import time

async def fetch_data(data_id: int) -> str:
    print(f'\t[Task fetch_data_{data_id}] Starting data fetch for ID {data_id}')
    await asyncio.sleep(2)  # Simulate I/O bound operation
    print(f'\t[Task fetch_data_{data_id}] Finished data fetch for ID {data_id}')
    return f'Data-{data_id}'

async def compute_heavy_task(task_id: int) -> int:
    print(f'\t[Task compute_heavy_task_{task_id}] Starting heavy computation for task {task_id}')
    await asyncio.sleep(3)  # Simulate CPU-bound operation (or just a longer delay)
    print(f'\t[Task compute_heavy_task_{task_id}] Finished heavy computation for task {task_id}')
    return task_id * 100

async def main_concurrent() -> None:
    start_time = time.time()
    print('Starting concurrent tasks...')

    # Use asyncio.gather to run tasks concurrently
    # The results will be returned in the order the coroutines are passed
    results = await asyncio.gather(
        fetch_data(1),
        compute_heavy_task(1),
        fetch_data(2)
    )

    print(f'All concurrent tasks finished in {time.time() - start_time:.2f} seconds.')
    print(f'Results: {results}')

# Run the main concurrent function directly in Colab
await main_concurrent()
```

### Explanation of Concurrent Execution:

In the output, you'll observe that the print statements from different tasks (`fetch_data` and `compute_heavy_task`) are interleaved. This indicates that they are indeed running concurrently rather than sequentially.

*   **Sequential Execution:** If these tasks ran one after another, the total time would be approximately `2 seconds (fetch_data_1) + 3 seconds (compute_heavy_task_1) + 2 seconds (fetch_data_2) = 7 seconds`.
*   **Concurrent Execution with `asyncio.gather`:** Since `fetch_data(1)` and `fetch_data(2)` both have a 2-second delay and `compute_heavy_task(1)` has a 3-second delay, the total execution time will be dominated by the longest running task (3 seconds for `compute_heavy_task_1`), plus a small overhead. You should see the total time being around 3 seconds, demonstrating the power of concurrency.

In Jupyter notebooks (including Colab), there's typically an event loop already running.
```python
# basic_threading.py
import threading
import time

def download_file(url, filename):
    """Download a single file"""
    thread_name = threading.current_thread().name
    print(f"[{thread_name}] Starting: {filename}")
    time.sleep(2)  # Simulate download time
    print(f"[{thread_name}] ✓ Completed: {filename}")
    return filename

def threaded_downloads():
    """Download files using multiple threads"""
    files = [
        ("https://example.com/video.mp4", "video.mp4"),
        ("https://example.com/document.pdf", "document.pdf"),
        ("https://example.com/music.mp3", "music.mp3")
    ]

    print("=== Multi-Threaded Downloads ===")
    start_time = time.time()

    threads = []

    # Step 1: Create threads
    for url, filename in files:
        thread = threading.Thread(
            target=download_file,
            args=(url, filename),
            name=f"Downloader-{filename.split('.')[0]}"
        )
        threads.append(thread)

    # Step 2: Start all threads
    for thread in threads:
        thread.start()

    # Step 3: Wait for all threads to complete
    for thread in threads:
        thread.join()

    total_time = time.time() - start_time
    print(f"Threaded time: {total_time:.1f} seconds")

if __name__ == "__main__":
    threaded_downloads()
```

```python
#subclassing thread class
import threading
import time
import random

class DownloadThread(threading.Thread):
    """Custom thread class for downloading files with built-in retry logic"""

    def __init__(self, url, filename, max_retries=3):
        super().__init__(name=f"Downloader-{filename.split('.')[0]}")
        self.url = url
        self.filename = filename
        self.max_retries = max_retries
        self.result = None
        self.download_time = None
        self.attempts = 0

    def run(self):
        """Called when thread.start() runs"""
        print(f"[{self.name}] Starting download: {self.filename}")
        start_time = time.time()

        for attempt in range(1, self.max_retries + 1):
            self.attempts = attempt
            try:
                print(f"[{self.name}] Attempt {attempt} for {self.filename}")
                time.sleep(2)  # Simulate download

                if attempt == 1 and random.random() < 0.2:  # 20% chance of failure
                    raise Exception("Network timeout")

                self.download_time = time.time() - start_time
                self.result = "success"
                print(f"[{self.name}] {self.filename} downloaded in {self.download_time:.1f}s")
                return
            except Exception as e:
                print(f"[{self.name}] Attempt {attempt} failed: {e}")
                if attempt < self.max_retries:
                    time.sleep(0.5)  # brief pause before retry
                else:
                    self.result = "failed"
                    self.download_time = time.time() - start_time
                    print(f"[{self.name}] Permanently failed after {attempt} attempts")

def demonstrate_custom_threads():
    files = [
        ("https://example.com/video.mp4", "video.mp4"),
        ("https://example.com/document.pdf", "document.pdf"),
        ("https://example.com/music.mp3", "music.mp3")
    ]

    print("=== Custom Thread Class Downloads ===")
    download_threads = [DownloadThread(url, filename) for url, filename in files]

    start_time = time.time()
    for t in download_threads: t.start()
    for t in download_threads: t.join()
    total_time = time.time() - start_time

    print("\n=== Download Results ===")
    successful = 0
    for t in download_threads:
        status = "Success" if t.result == "success" else "Failed"
        print(f"{status} - {t.filename}: {t.result} ({t.attempts} attempts, {t.download_time:.1f}s)")
        if t.result == "success": successful += 1

    print(f"Total time: {total_time:.1f} seconds")
    print(f"Success rate: {successful}/{len(download_threads)}")

if __name__ == "__main__":
    demonstrate_custom_threads()
```

```python
#pass arguments to threads
import threading, time, random

def download_with_options(file_id, filename, size_mb=1, priority="normal", max_retries=3):
    """Simulate downloading a file with various options and retry logic"""
    thread = threading.current_thread().name
    print(f"[{thread}] Download {file_id}: {filename} ({size_mb}MB, {priority})")

    # Calculate download time based on size and priority
    base_time = size_mb * 0.3
    if priority == "high":
        base_time *= 0.7  # High priority downloads faster
    if priority == "low":
        base_time *= 1.3  # Low priority downloads slower

    # Retry logic with simulated failures
    for attempt in range(max_retries):
        try:
            print(f"[{thread}] Attempt {attempt+1} for {filename}")
            time.sleep(base_time)  # Simulate download time

            # Simulate random network failures on first attempt
            if attempt == 0 and random.random() < 0.2:
                raise Exception("Network timeout")

            print(f"[{thread}] Downloaded {filename} successfully")
            return
        except Exception as e:
            print(f"[{thread}] Failed attempt {attempt+1}: {e}")
            time.sleep(0.5)  # Wait before retry

def demonstrate_safe():
    """Show how immutable arguments are safely passed to threads"""
    print("=== Safe argument passing ===")

    # Create threads with different argument combinations
    t1 = threading.Thread(target=download_with_options, args=(1, "video.mp4"))
    t2 = threading.Thread(
        target=download_with_options,
        args=(2, "document.pdf"),
        kwargs={"size_mb": 5, "priority": "high"}
    )
    t3 = threading.Thread(
        target=download_with_options,
        args=(3, "music.mp3"),
        kwargs={"priority": "low"}
    )

    # Start all threads
    for t in [t1, t2, t3]:
        t.start()

    # Wait for all threads to complete
    for t in [t1, t2, t3]:
        t.join()

if __name__ == "__main__":
    demonstrate_safe()
```

### Data Class = A special kind of class that's designed mostly for holding data without writing a lot of the boilerplate code for regular classes.
### They automatically generate: _init__, __repr__, __eq_

```python
class Person:

  def __init__(self,name,age):
    self.name=name
    self.age=age
    self.is_alive=True

  def __repr__(self):
    return f"(name={self.name}, age={self.age}, is_alive={self.is_alive})"

  def __eq__(self, other):
    if not isinstance(other, Person):
      return NotImplemented
    return self.name == other.name and self.age == other.age

Person1 = Person("Siddhi", 22)
Person2 = Person("Siddhi", 22)

print(Person1)
print(Person2)
print(Person1 == Person2)

```
