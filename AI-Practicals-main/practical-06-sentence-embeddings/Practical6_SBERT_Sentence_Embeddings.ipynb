- t-SNE visualization of embeddings

## Step 1: Import Libraries

```
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.manifold import TSNE

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

## Step 2: Create Sentence Dataset

We create 20 sentences grouped into 4 semantic topics:
- Sports
- Technology
- Food
- Travel


```
sentences = [

    # Sports
    "The football team won the championship.",
    "Cricket is one of the most popular sports in India.",
    "The player scored a brilliant goal in the final match.",
    "Basketball requires speed and teamwork.",
    "The athlete trained hard for the Olympics.",

    # Technology
    "Artificial intelligence is transforming industries.",
    "Python is widely used for machine learning projects.",
    "Cybersecurity is important for protecting data.",
    "Cloud computing enables scalable applications.",
    "Data science combines statistics and programming.",

    # Food
    "Pizza tastes delicious with extra cheese.",
    "Indian curries are rich in flavor and spices.",
    "Fresh fruits are important for a healthy diet.",
    "Chocolate ice cream is my favorite dessert.",
    "Pasta is commonly served in Italian restaurants.",

    # Travel
    "The mountains look beautiful during winter.",
    "I enjoy visiting beaches during summer vacations.",
    "Tourists love exploring historical monuments.",
    "The flight to Paris was comfortable and relaxing.",
    "Traveling helps people experience different cultures."
]

topics = [
    "Sports", "Sports", "Sports", "Sports", "Sports",
    "Technology", "Technology", "Technology", "Technology", "Technology",
    "Food", "Food", "Food", "Food", "Food",
    "Travel", "Travel", "Travel", "Travel", "Travel"
]

print(f"Total Sentences: {len(sentences)}")
```

## Step 3: Load SBERT Model

```
model = SentenceTransformer('all-MiniLM-L6-v2')

print("SBERT model loaded successfully!")
```

## Step 4: Generate Sentence Embeddings

```
embeddings = model.encode(sentences)

print("Embedding Shape:", embeddings.shape)
```

## Step 5: Compute Pairwise Cosine Similarity

```
similarity_matrix = cosine_similarity(embeddings)

# Convert to DataFrame for better visualization
similarity_df = pd.DataFrame(
    similarity_matrix,
    columns=[f"S{i+1}" for i in range(len(sentences))],
    index=[f"S{i+1}" for i in range(len(sentences))]
)

similarity_df.head()
```

## Step 6: Visualize Similarity Matrix

```
plt.figure(figsize=(12,10))

sns.heatmap(similarity_matrix, cmap='coolwarm')

plt.title("Cosine Similarity Matrix")
plt.xlabel("Sentence Index")
plt.ylabel("Sentence Index")

plt.show()
```

## Step 7: Query Sentence Matching

We find the most semantically similar sentence for a given query.


```
query = "Machine learning and artificial intelligence are changing technology."

# Encode query
query_embedding = model.encode([query])

# Compute similarity scores
scores = cosine_similarity(query_embedding, embeddings)[0]

# Find best matching sentence
best_index = np.argmax(scores)

print("Query Sentence:")
print(query)

print("\nBest Matching Sentence:")
print(sentences[best_index])

print("\nSimilarity Score:")
print(scores[best_index])
```

## Step 8: Apply t-SNE Dimensionality Reduction

SBERT embeddings are high-dimensional (384 dimensions).
t-SNE reduces them to 2 dimensions for visualization.


```
tsne = TSNE(
    n_components=2,
    random_state=42,
    perplexity=5
)

reduced_embeddings = tsne.fit_transform(embeddings)

print("Reduced Embedding Shape:", reduced_embeddings.shape)
```

## Step 9: Visualize Sentence Embeddings using t-SNE

```
plt.figure(figsize=(12,8))

sns.scatterplot(
    x=reduced_embeddings[:,0],
    y=reduced_embeddings[:,1],
    hue=topics,
    s=120
)

# Annotate points
for i, sentence in enumerate(sentences):
    plt.text(
        reduced_embeddings[i,0] + 0.3,
        reduced_embeddings[i,1] + 0.3,
        str(i+1),
        fontsize=9
    )

plt.title("t-SNE Visualization of SBERT Sentence Embeddings")
plt.xlabel("t-SNE Dimension 1")
plt.ylabel("t-SNE Dimension 2")

plt.legend(title="Topics")

plt.show()
```

## Step 10: Observations

### Observations:
- Sentences belonging to similar topics appear closer together.
- Technology-related sentences form meaningful clusters.
- Food and travel sentences are grouped semantically.
- SBERT captures contextual meaning effectively.

### Conclusion:
The practical demonstrated that SBERT sentence embeddings successfully capture semantic similarity between sentences. Cosine similarity identified related sentences accurately, while t-SNE visualization revealed meaningful topic-based clustering.
